```python
import os
import shutil
import kagglehub
import pandas as pd

# 1. Download raw dataset
path = kagglehub.dataset_download("tmthyjames/nashville-housing-data")
csv_files = [f for f in os.listdir(path) if f.lower().endswith(".csv")]
raw_csv_path = os.path.join(path, csv_files[0])
print("Raw CSV:", raw_csv_path)

# 2. Fresh clone of your data sanitizer repo
repo_dir = "/content/universal-data-sanitizer"
if os.path.exists(repo_dir):
    shutil.rmtree(repo_dir)

!git clone https://github.com/Farhan1Ahmed1Molla/universal-data-sanitizer.git {repo_dir}

# 3. Run data_sanitizer.py to produce cleaned CSV
!python {repo_dir}/data_sanitizer.py "{raw_csv_path}"

# 4. Locate the generated file in /content/ and save a clean copy
base_name = os.path.splitext(os.path.basename(raw_csv_path))[0]
generated_clean = f"/content/{base_name}_clean.csv"
clean_path = "/content/nashville_housing_clean.csv"

if os.path.exists(generated_clean):
    shutil.copyfile(generated_clean, clean_path)
elif not os.path.exists(clean_path):
    raise FileNotFoundError(f"Expected cleaned file at {generated_clean}")

# 5. Verify cleaned output
df_clean = pd.read_csv(clean_path)
print("Saved to:", clean_path)
print("Clean shape:", df_clean.shape)
print("Clean columns:", df_clean.columns.tolist())
```

```python
df_work = df_clean.copy()
print("Working copy created:", df_work.shape)
```

```python
print("Head:")
print(df_work[["unnamed_01", "unnamed_0"]].head(5))

print("Tail:")
print(df_work[["unnamed_01", "unnamed_0"]].tail(5))

for col in ["unnamed_01", "unnamed_0"]:
    expected = pd.Series(range(len(df_work)), index=df_work.index)
    is_sequential_0 = df_work[col].reset_index(drop=True).equals(expected)
    print(f"{col}: nunique{df_work[col].nunique()}, min={df_work[col].min()}, max={df_work[col].max()}, exact 0..N-1={is_sequential_0}")
```

```python
df_work = df_work.drop(columns=["unnamed_01", "unnamed_0"])

print("New shape after dropping index columns:", df_work.shape)

print("First 5 columns now", df_work.columns[:5].tolist())
```

```python
print("--- Missing counts ---")
print(df_work[["property_address", "address", "property_city", "city"]].isna().sum())

both_addr = df_work["property_address"].notna() & df_work["address"].notna()
print("\nBoth Addresses non-null:", both_addr.sum())
print("Exact match when both non-null:", (df_work.loc[both_addr, "property_address"] == df_work.loc[both_addr, "address"]).sum())


both_city = df_work["property_city"].notna() & df_work["city"].notna()
print("\nBoth Cities non-null:", both_city.sum())
print("Exact match when both non-null:", (df_work.loc[both_city, "property_city"] == df_work.loc[both_city, "city"]).sum())

print("\n--- Sample rows where both address are non-null ---")
display(df_work.loc[both_addr, ["parcel_id", "property_address", "address", "property_city", "city"]])
```

```python
both_addr = df_work["property_address"].notna() & df_work["address"].notna()
mismatch_addr = df_work.loc[both_addr & (df_work["property_address"] != df_work["address"]),
                            ["parcel_id", "property_address", "address", "owner_name"]]
print("Non-null address mismatches:", len(mismatch_addr))
display(mismatch_addr.head(10))

fills_missing_prop_addr = df_work["property_address"].isna() & df_work["address"].notna()
print("Rows where property address is Nan and address is not Nan:", fills_missing_prop_addr.sum())



```

```python
fill_mask = df_work["property_address"].isna() & df_work["address"].notna()
print("--- Rows where property address is Nan and address is NOT NaN ---")
display(df_work.loc[fill_mask, ["parcel_id", "property_address", "address", "city", "owner_name"]])

both_city = df_work["property_city"].notna() & df_work["city"].notna()
city_diff = df_work.loc[both_city & (df_work["property_city"] != df_work["city"]),
                        ["parcel_id", "property_address", "address", "property_city", "city"]]

print("\n--- Rows where both cities are non-null and differ ---")
display(city_diff)
```

```python
df_work = df_work.drop(columns=["address", "city"])
print("New shape after dropping address and city:", df_work.shape)
print("Remaining columns count:", len(df_work.columns))
```

```python
for col in ["sold_as_vacant", "multiple_parcels_involved_in_sale", "property_city"]:
  print(f"\n--- {col} ---")
  print(df_work[col].value_counts(dropna=False))
```

```python
missing_city = df_work[df_work["property_city"].isna()]
print("Missing property_city rows:", len(missing_city))
display(missing_city[["parcel_id", "property_address", "property_city", "land_use", "sale_price"]].head(10))
```

```python
known_city_by_parcel = (
    df_work.dropna(subset=["property_city"])
    .groupby("parcel_id")["property_city"]
    .first()
)

fillable = missing_city["parcel_id"].map(known_city_by_parcel)
print("Missing property_city rows with a known city on another row of same parcel_id:", fillable.notna().sum())
```

```python
print("sale_date dtype:", df_work["sale_date"].dtype)
print("Missing sale_date:", df_work["sale_date"].isna().sum())
print("\nFirst 10 sale_date values:")
print(df_work["sale_date"].head(10).tolist())
print("\nSample 10 random non-null sale_date values:")
print(df_work["sale_date"].dropna().sample(10, random_state=42).tolist())
```

```python
parsed_sale_date = pd.to_datetime(df_work["sale_date"], format="%Y-%m-%d", errors="coerce")
print("Failed parses (NaT count):", parsed_sale_date.isna().sum())
print("Min date:", parsed_sale_date.min(), "| Max date:", parsed_sale_date.max())
```

```python
df_work["sale_date"] = parsed_sale_date
print("Updated sale_date dtype:", df_work["sale_date"].dtype)
```

```python
df_work["sale_date"] = parsed_sale_date

num_cols = [
    "sale_price", "land_value", "building_value", "total_value",
    "acreage", "finished_area", "year_built", "bedrooms", "full_bath", "half_bath"
]
display(df_work[num_cols].describe().T[["count", "min", "50%", "max"]])
```

```python
low_sales = df_work.nsmallest(10, "sale_price")[
    ["parcel_id", "sale_date", "sale_price", "land_use", "sold_as_vacant", "total_value"]
]
display(low_sales)
```

```python
for cutoff in [100, 1000, 10000]:
    count = (df_work["sale_price"] < cutoff).sum()
    print(f"Rows with sale_price < {cutoff}: {count}")
```

```python
print("Shape:", df_work.shape)
print(df_work.dtypes)
```

```python
missing_summary = (
    df_work.isna()
    .sum()
    .to_frame("missing_count")
    .assign(missing_pct=lambda d: (d["missing_count"] / len(df_work) * 100).round(2))
    .sort_values("missing_count", ascending=False)
)
display(missing_summary)
```

```python
yearly_summary = (
    df_work.assign(sale_year=df_work["sale_date"].dt.year)
    .groupby("sale_year")["sale_price"]
    .agg(sales_count="count", median_price="median", mean_price="mean")
)
display(yearly_summary)
```

```python
text_cols = ["land_use", "property_city", "sold_as_vacant", "multiple_parcels_involved_in_sale"]
for col in text_cols:
    s = df_work[col].dropna().astype(str)
    dirty_count = (s != s.str.strip()).sum()
    print(f"{col}: {dirty_count} values with leading/trailing whitespace")
```

```python
df_work.to_csv("real_estate_cleaned.csv", index=False)
print("Saved real_estate_cleaned.csv with shape:", df_work.shape)
```

```python
print("--- By sold_as_vacant ---")
display(
    df_work.groupby("sold_as_vacant")["sale_price"]
    .agg(sales_count="count", median_price="median", mean_price="mean")
)

print("\n--- Top 8 land_use by sales_count ---")
top_land_use = df_work["land_use"].value_counts().head(8).index
display(
    df_work[df_work["land_use"].isin(top_land_use)]
    .groupby("land_use")["sale_price"]
    .agg(sales_count="count", median_price="median", mean_price="mean")
    .sort_values("sales_count", ascending=False)
)
```

```python
print("Unique land_use count:", df_work["land_use"].nunique())
display(df_work["land_use"].value_counts())
```

```python
land_use_fixes = {
    "VACANT RES LAND": "VACANT RESIDENTIAL LAND",
    "VACANT RESIENTIAL LAND": "VACANT RESIDENTIAL LAND",
    "GREENBELT/RES\r\nGRRENBELT/RES": "GREENBELT/RES",
}

df_work["land_use"] = df_work["land_use"].replace(land_use_fixes)
print("Updated unique land_use count:", df_work["land_use"].nunique())
display(df_work["land_use"].value_counts().head(8))

df_work.to_csv("real_estate_cleaned.csv", index=False)
print("Updated real_estate_cleaned.csv saved.")
```
